```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
import numpy as np
import pandas as pd

x = np.array([1.3, 1.5, 1.7, 2.5, 2.7, 2.5, 3.2, 3.0, 3.5, 4.2, 2.5])

y = np.array([
    49000, 56000, 37700, np.nan, 49800,
    56000, 60100, 55400, 57100, 65200, 57600
])

df = pd.DataFrame({
    "Experience": x,
    "Salary": y
})

print(df)
```

```python
plt.plot(x, y, marker="D")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Original Data - Experience vs Salary")
plt.grid()
plt.show()
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df
```

```python
df = df.drop_duplicates()

df
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
plt.plot(x, y, marker="D")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Original Data - Experience vs Salary")
plt.grid()
plt.show()
```

```python
x = df[["Experience"]]
y = df["Salary"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=1
)

print("X Train Shape:", x_train.shape)
print("X Test Shape:", x_test.shape)
print("Y Train Shape:", y_train.shape)
print("Y Test Shape:", y_test.shape)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)

print("Prediction value:", y_pred)
```

```python
r2 = r2_score(y_test, y_pred)

print("R2 Score:", r2)
```

```python
experience = np.array([[5]])

salary = model.predict(experience)

print("Predicted Salary for 5 Years Experience:", salary[0])
```

```python
plt.plot(x, y, label="Actual Salary")
plt.plot(x, model.predict(x), label="Regression Line")

plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear Regression - Experience vs Salary")

plt.legend()
plt.grid()
plt.show()
```
